Imagem Digital = armazenada e enxrgeada como sendo uma matriz

Cada posição da matriz é um pixel = picture element.
Cada pixel é a menor unidade de informação que pode guardar uma imagem
RGB - red, green, blue.
Sistema de cores mais utilizado nos sitemas operacionais modernos.
Sistema LAB, HSV...

O RGB é formado assim - cada pixel possui 1 byte de info - 8 bits de informaççao. 1 bit é uma vari´vael de 0 ou 1.
01011011

2^8 = 256 tons de cinza.
0 - preto absoluto
255 - branco absoluto.

Matriz tridimensional.

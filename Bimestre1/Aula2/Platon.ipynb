```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
img_bgr = cv2.imread('platon.jpg') #fica assim pq o padrão do open cv é bgr, ou seja, o padrão das cores fica invertido
img_bgr
```

```python
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
img_rgb
```

```python
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
img_hsv
```

```python
plt.figure(figsize=(5, 3)) #criar um frame de imagem, um quadro de tam 500 px x 300px
plt.imshow(img_rgb)
plt.title('img_rgb')
plt.axis('off')

plt.show()
```

```python
#Converter para cinza
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
img_gray
```

```python
#Matriz 2 x 2
fig, ax = plt.subplots(ncols=2, nrows=2, figsize=(6, 4))

#1 imagem
ax[0, 0].imshow(img_rgb)
ax[0, 0].set_title('RGB')
ax[0, 0].axis('off')

#2 imagem
ax[0, 1].imshow(img_bgr)
ax[0, 1].set_title('BGR')
ax[0, 1].axis('off')

#3 imagem
ax[1, 0].imshow(img_hsv)
ax[1, 0].set_title('HSV')
ax[1, 0].axis('off')

#4 imagem
ax[1, 1].imshow(img_gray, cmap='gray') #o pyplot converte imagens que estão de uma só cor em uma escala, colormap so funciona para imagens cinzas, POSSUI 1 ÚNICO CANAL
ax[1, 1].set_title('GRAY')
ax[1, 1].axis('off')

plt.tight_layout()

plt.savefig('platon_convertido.png')
plt.show()
```
